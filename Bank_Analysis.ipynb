```python
categorical_columns = [
    'Age_T',
    'Home Ownership',
    'Loan Status',
    'Product Code',
    'Grade',
    'Sub Grade',
    'Purpose Category',
    'Region Name',
    'Verification Status',
    'Application Type'
]

for col in categorical_columns:
    if col in df.columns:
        df[col] = df[col].fillna('Unknown')
```

```python
numerical_columns = [
    'Loan Amount',
    'Funded Amount',
    'Int Rate',
    'Total Pymnt',
    'Total Rec Prncp',
    'Recoveries'
]

for col in numerical_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
date_columns = [
    'Dateof Birth',
    'Closed Date',
    'Disbursement Date',
    'Loan Transferdate',
    'NextMeetingDate'
]

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(
            df[col],
            errors='coerce'
        )
```

```python
print("Before:", df.shape)

df.drop_duplicates(inplace=True)

print("After:", df.shape)
```

```python
df['Disbursement Year'] = (
    df['Disbursement Date'].dt.year
)
```

```python
df['Disbursement Month'] = (
    df['Disbursement Date'].dt.month_name()
)
```

```python
df['Loan Category'] = pd.cut(
    df['Loan Amount'],
    bins=[-np.inf, 50000, 100000, 500000, np.inf],
    labels=[
        'Small Loan',
        'Medium Loan',
        'Large Loan',
        'Very Large Loan'
    ]
)
```

```python
df['Outstanding Principal'] = (
    df['Funded Amount'] - df['Total Rec Prncp']
)
```

```python
df['Recovery Percentage'] = np.where(
    df['Funded Amount'] > 0,
    (df['Recoveries'] / df['Funded Amount']) * 100,
    np.nan
)
```

## Exploratory Data Analysis (EDA)

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x='Loan Status',
    order=df['Loan Status'].value_counts().index
)

plt.title('Loan Status Distribution')
plt.xlabel('Loan Status')
plt.ylabel('Number of Loans')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
bank_loan = (
    df.groupby('Bank Name')['Loan Amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,5))

bank_loan.plot(kind='bar')

plt.title('Top 10 Banks by Loan Amount')
plt.xlabel('Bank Name')
plt.ylabel('Total Loan Amount')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
state_loan = (
    df.groupby('State Name')['Loan Amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,5))

state_loan.plot(kind='bar')

plt.title('Top 10 States by Loan Amount')
plt.xlabel('State')
plt.ylabel('Loan Amount')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
purpose = df['Purpose Category'].value_counts().head(10)

plt.figure(figsize=(10,5))

purpose.plot(kind='barh')

plt.title('Top 10 Loan Purposes')
plt.xlabel('Number of Loans')
plt.ylabel('Purpose Category')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    df['Loan Amount'].dropna(),
    bins=30,
    kde=True
)

plt.title('Loan Amount Distribution')
plt.xlabel('Loan Amount')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.scatterplot(
    data=df,
    x='Loan Amount',
    y='Funded Amount',
    alpha=0.5
)

plt.title('Loan Amount vs Funded Amount')
plt.xlabel('Loan Amount')
plt.ylabel('Funded Amount')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    df['Int Rate'].dropna(),
    bins=30,
    kde=True
)

plt.title('Interest Rate Distribution')
plt.xlabel('Interest Rate')
plt.ylabel('Number of Loans')

plt.tight_layout()
plt.show()
```

```python
monthly_loan = (
    df.groupby('Disbursement Month')['Funded Amount']
    .sum()
)

month_order = [
    'January', 'February', 'March',
    'April', 'May', 'June',
    'July', 'August', 'September',
    'October', 'November', 'December'
]

monthly_loan = monthly_loan.reindex(month_order)

plt.figure(figsize=(10,5))

monthly_loan.plot(kind='line', marker='o')

plt.title('Monthly Loan Disbursement Trend')
plt.xlabel('Month')
plt.ylabel('Funded Amount')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
delinquent = df['Is Delinquent Loan'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    delinquent.values,
    labels=delinquent.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Delinquent Loan Distribution')

plt.show()
```

```python
top_customers = (
    df.groupby('Client Name')['Loan Amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_customers)
```

```python
branch_performance = (
    df.groupby('Branch Name')['Funded Amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(branch_performance)
```
